1. What is nf-core?

nf-core is a collaborative community that intends to enhance the reproducability of bioinformatics research. This is done by providing standardised sets of guidelines, and templates for building bioinfomatics pipelines. 

2. How many pipelines are there currently in nf-core?

157 according to the nf-core Website (https://nf-co.re/pipelines/)

3. Are there any non-bioinformatic pipelines in nf-core?

Yes, there are non-bioinformatics pipelines in nf-core. For example there are spinningjenny(which simulates the first industrial revolution),meerpipe(which is an astronomy pipeline for pulsa data), datasync(which is a pipeline for copying files and directories between stroage spaces).

4. Let's go back a couple of steps. What is a pipeline and what do we use it for?

A pipeline is at its core a strict set of guidelines/rules to move (in computational context usually data) soemthing through a fixed sequence of steps. This process is usually automated. Data pipelines for example tend to take raw data and automatically move the data through a fixed set of processing steps, until a (hopefully) useable output is archived. 

5. Why do you think nf-core adheres to strict guidelines?

As one of the goals of nf-core is to ensure better reproducability, these strict guidelines are a necassity, as they ensure order and stable releases, which in turn ensure the reproducabilty.

6. What are the main features of nf-core pipelines?

- Extensive Documentation 
- stable releases
- open source
- continous integration testing
- run on most laptops, HPCs and clouds 
- Pipeline dependencies are automatically handled
- parallelization, orchastration, infrastructure config

### What is differential abundance analysis?

Differential Abundance analysis aims to find differences in abundance between samples or conditions. Looks into expression differences in genes.

Give the most important plots from the report:

The Differential Abundance Analysis, as the name implies, tries to find the differences between samples. Therefore the differential plots are the most interesting. Though of course we always need to look at differences in context. So understanding the data we have is key. One important plot to look at would be the PCA plots. The PCA plots allow a quick glance at our data and early on already gives us rather intersting clues about the relationship between different samples. The PCA plot might also already indicate outliers or other experimental errors (though to confirm these further exploration is necessary). A PCA plot reduces features down into few dimensions, and shows the dimensions or directions of the most variance. So here it is fairly interesting to look at the first few Principal components to check how much variance is covered by them. Usually most variance should be explained by the first two to three principal components. If that is not the case in all prinicpal components are very similar, this would indicate that the groups are also rather similar and dont ahve a strong variance in any direction (unfortunatly this also makes the pca a bit useless). 
Furthermore samples of the same kind (no change, no different taxa, same experimental condition) should group together. So the PCA can already indicate to us for one, that there might be outliers or experimental errors (wrong label, wrong pipetting,e tc.), and it can show us differentially expressed clusteres. 
But what can we see here: In the PCA the mCherry appears to have a higher variance between its own samle than when compared to the hND6 sample. Which makes it hard to say that the treatmant has an important effect.
